```python
import os
import sys
from pathlib import Path
from ai2thor.controller import Controller
from src.simulation.thor_env import ThorEnv, ACTION_LIST


def go_to_root(root_name="resaction-nav"):
    p = Path.cwd()
    for parent in [p] + list(p.parents):
        if parent.name == root_name:
            os.chdir(parent)
            sys.path.append(str(parent))
            return parent
    raise RuntimeError(f"Root '{root_name}' not found")

ROOT = go_to_root()
print("cwd:", Path.cwd())
```

```python
import numpy as np
from thor_env import ThorEnv
import torch.nn as nn
import random
```

```python
env = ThorEnv(device='cpu')
```

```python
class test_model(nn.Module):
    def __init__(self):
        super().__init__()
        self.name = 'test'

    def forward(self, obs):
        choice = random.choice([
            0,
            1,
            2,
            3,
            4,
            5,
            6,
            7,
            8,
        ])
        return choice
```

```python
obs = env.reset(scene='FloorPlan21')
```

```python
model = test_model()
obs = env._compute_obs()
rollout = []
for _ in range(5):
    action_idx = model(obs)
    obs, reward, terminated, truncated, info = env.step(action_idx)
    rollout.append((env.action_list[action_idx], reward, terminated, truncated))
    if terminated or truncated:
        break

rollout
```
